# Steady-state Laminar FLames

## Model equations

The Euler equations in 1D (including reactions and thermal diffusion) are:

$$
\frac{\partial \rho}{\partial t} + \frac{\partial (\rho u)}{\partial x} &= 0 \\
\frac{\partial \rho u}{\partial t} + \frac{\partial (\rho u^2 + p)}{\partial x} &= 0 \\
\frac{\partial \rho E}{\partial t} + \frac{\partial (\rho u E + u p)}{\partial x} &= \rho \epsilon + \frac{\partial}{\partial x} \sigma \frac{\partial T}{\partial x} \\
\frac{\partial \rho X_k}{\partial t} + \frac{\partial (\rho u X_k)}{\partial x} &= \rho \dot{\omega}_k
$$

For flames, it is preferable to work in terms of enthalpy, so we replace the energy equation with:

$$
\frac{\partial \rho h}{\partial t} + \frac{\partial (\rho u h)}{\partial x} = \frac{Dp}{Dt} + \rho \epsilon + \frac{\partial}{\partial x} \sigma \frac{\partial T}{\partial x}$$

where $Dp/Dt$ is the Lagrangian derivative of pressure.

### Steady state / constant pressure approximations

Now we assume that we are in a steady state, so all the time derivatives vanish.  We also assume that the pressure is constant (since flames are subsonic, and pressure is constant to $\mathcal{O}(M^2)$) so $Dp/Dt = 0$.

This gives:

* mass conservation:

  $$\frac{d (\rho u)}{dx} = 0$$

  which we write as $m = \rho u = \text{constant}$, where $m$ is the mass flux.

* momentum conservation:

  $$\frac{d (\rho u^2 + p)}{dx} = 0$$

  we can take this as $m^2/\rho + p = \text{constant}$, which would give us a prescription for $\rho(x)$ throughout the domain,
  but instead we will replace this with $p = p_0 = \text{constant}$, and then compute $\rho$ as needed from:

  $$\rho = \rho(p_0, T, X_k)$$

* energy / enthalpy conservation:

  $$\frac{d (\rho u h)}{dx} = \rho \epsilon + \frac{d}{dx} \sigma \frac{dT}{dx}$$

  defining the heat flux as

  $$q = - \sigma \frac{dT}{dx}$$

  we have:

  $$m \frac{dh}{dx} = \rho \epsilon - \frac{dq}{dx}$$

  and we can eliminate the enthalpy gradient via the chain rule:

  $$\frac{dh}{dx} = \underbrace{\left . \frac{\partial h}{\partial T} \right |_{p,X_k}}_{\equiv c_p} \frac{dT}{dx} + \sum_k \underbrace{\left . \frac{\partial h}{\partial X_k} \right |_{T,p,X_{j\ne k}}}_{\equiv \xi_k} \frac{dX_k}{dx}$$

  where we identified the specific heat and used the fact that $dp/dx = 0$.  Now using:

  $$\frac{dT}{dx} &= -\frac{q}{\sigma} \\
    \frac{dX_k}{dx} &= \frac{\rho}{m} \dot{\omega}_k$$

   (see the next bullet for species) we get:

  $$\frac{dq}{dx} = \frac{mc_p q}{\sigma} + \rho \left (\epsilon - \sum_k \xi_k \dot{\omega}_k \right )$$

* species conservation

  We have:

  $$\frac{d(\rho u X_k)}{dx} = \rho \dot{\omega}_k$$

  which we can write as:

  $$\frac{dX_k}{dx} = \frac{\rho}{m} \dot{\omega}_k$$  

### Final equations

Our final set of equations are:

$$
\frac{dT}{dx} &= -\frac{q}{\sigma} \\
\frac{dq}{dx} &= \frac{m c_p q}{\sigma} + \rho \left ( \epsilon - \sum_k \xi_k \dot{\omega}_k \right ) \\
\frac{dX_k}{dx} &= \frac{\rho}{m} \dot{\omega}_k$$

together with $p = p_0$, $m = \text{constant}$.

Note that Chamulak et al. 2007 say in a footnote that they ignore the $\xi_k$ terms.

In total our solution state is ${\bf y} = (T, q, X_k)$, which is $N + 2$ unknowns (with $N$ the number of species in the network).

## Boundary conditions

This is a system of first order ODEs, so we need a boundary condition for each $N + 2$ quantity + an additional one for $m$.

### Upstream boundary conditions

Ahead of the flame, we want:

* $T(-\infty) = T_f$
* $X_k(-\infty) = X_{k,f}$
* $q(-\infty) = 0$ : no heat flux far ahead of the flame

where the $f$ subscript indicates "fuel".  

We also should require that there is no burning, $\dot{\omega}_k(\rho_f, T_f, X_{k,f}) = 0$, or that it is very small.

Practically, we are on a finite domain, so if we are evolving on $[-L, L]$ instead of $[-\infty,\infty]$ we expect there to be some temperature gradient, so we can take:

$$T(-L) = T_f + \delta T$$

This serves an important purpose---it fixes the position of the flame in the domain (since the solution on the infinite domain is translationally invariant).
We have to choose $\delta T$ for our problem (Chamulak et al. 2007 just state they
they set $dT/dx$ to an arbitrary small value).

Since this implies a heat flux, we need to prescribe a finite boundary value for $q$.  Ignoring nuclear reactions, we can start with:

$$\frac{dq}{dx} = - m \frac{dh}{dx}$$

and integrate from $[-\infty, -L]$ to get:

$$q(-L) = -m \left [ h(T_f + \delta T, p_0, X_{k,f}) - h(T_f, p_0, X_{k,f} )\right ]$$

where we use the EOS to evaluate the enthalpy.

### Downstream boundary conditions

Far behind the flame, the asymptotic condition is:

$$q(\infty) = 0$$

on our finite domain, we take:

$$q(L) = 0$$

This requires that all of the nuclear burning has ceased.  Practically, this means that we should not include weak reactions that have decays far from the burning region of the flame.  Chamulak et al. 2007 note that they do not include the weak reactions.

### Overconstrained?

Our mass fractions satisfy:

$$\sum_k X_k = 1$$

so only $N - 1$ mass fractions are independent.  This can cause the matrix involved
in the solution process to become singular, so an alternate approach is to eliminate one species (like the last) and use:

$$X_N = 1 - \sum_{k=1}^{N-1} X_k$$

This also drops the number of required boundary conditions to $N-2$.

## Consistency check

Starting with

$$m \frac{dh}{dx} = - \frac{dq}{dx} + \rho \epsilon$$

and integrating across the flame:

$$m (h_\text{ash} - h_f) = -[q(+L) - q(-L)] + \int_{-L}^{+L} \rho \epsilon dx$$

if the fluxes at the boundaries are small, then this is:

$$m (h_\text{ash} - h_f) =  \int_{-L}^{+L} \rho \epsilon dx$$

This says that the change in enthalpy across the flame is balanced by the nuclear energy release.

## Implementation

```python
import numpy as np
from scipy.integrate import solve_bvp
```

```python
def eos(p, T, X):
    """
    Evaluate EOS at fixed pressure.

    Parameters
    ----------
    p : float
        Pressure.
    T : float
        Temperature.
    X : ndarray
        Full mass fraction vector, shape (nspec,).

    Returns
    -------
    rho : float
        Density.
    h : float
        Specific enthalpy, excluding nuclear binding energy.
    cp : float
        Specific heat at constant pressure.
    h_X : ndarray
        Enthalpy composition derivatives at constant T and p,
        shape (nspec,).
    """
    raise NotImplementedError
```

```python
def conductivity(rho, T, X):
    """
    Return thermal conductivity in erg / (cm s K).
    """
    raise NotImplementedError
```

```python
def nuclear_rhs(rho, T, X):
    """
    Return instantaneous nuclear reaction rates.

    Returns
    -------
    Xdot : ndarray
        dX/dt from nuclear reactions, shape (nspec,), in 1/s.
    enuc : float
        Nuclear heating rate in erg / (g s).
        Include neutrino losses as appropriate.
    """
    raise NotImplementedError
```

```python
# ------------------------------------------------------------
# BVP problem
# ------------------------------------------------------------

class FlameBVP:

    def __init__(self, rho_fuel, T_fuel, X_fuel,
                 delta_T, L):

        self.rho_fuel = rho_fuel
        self.T_fuel = T_fuel
        self.X_fuel = np.asarray(X_fuel, dtype=float)

        self.nspec = len(X_fuel)
        self.nind = self.nspec - 1

        self.delta_T = delta_T
        self.L = L

        # The upstream EOS pressure must be supplied or
        # calculated from rho_fuel, T_fuel, X_fuel.
        self.p0 = self.fuel_pressure()

        # Upstream enthalpy
        _, self.h_fuel, _, _ = eos(
            self.p0, self.T_fuel, self.X_fuel
        )

    def fuel_pressure(self):
        """
        Implement the EOS inversion/evaluation needed to get
        p(rho_fuel, T_fuel, X_fuel).
        """
        raise NotImplementedError

    def full_composition(self, Xind):
        """
        Reconstruct the final species from normalization.

        Xind has shape (nspec - 1,).
        """
        X = np.empty(self.nspec)

        X[:-1] = Xind
        X[-1] = 1.0 - np.sum(Xind)

        return X

    def rhs(self, x, y, params):
        """
        Spatial ODEs.

        y has shape (nspec + 1, len(x)):
            y[0]  = T
            y[1]  = q
            y[2:] = independent mass fractions

        params[0] = mass flux m
        """

        m = params[0]

        dydx = np.empty_like(y)

        for j in range(x.size):

            T = y[0, j]
            q = y[1, j]

            X = self.full_composition(y[2:, j])

            rho, h, cp, h_X = eos(self.p0, T, X)

            kappa = conductivity(rho, T, X)

            Xdot, enuc = nuclear_rhs(rho, T, X)

            composition_term = np.dot(h_X, Xdot)

            dydx[0, j] = -q / kappa

            dydx[1, j] = (
                m * cp * q / kappa
                + rho * (enuc - composition_term)
            )

            dydx[2:, j] = rho * Xdot[:-1] / m

        return dydx

    def bc(self, ya, yb, params):
        """
        Boundary conditions.

        ya: state at x = -L
        yb: state at x = +L

        Returns nspec + 2 residuals.
        """

        m = params[0]

        T_left = ya[0]
        q_left = ya[1]

        X_left = self.full_composition(ya[2:])

        _, h_left, _, _ = eos(
            self.p0, T_left, X_left
        )

        residuals = np.empty(self.nspec + 2)

        # Upstream independent mass fractions
        residuals[:self.nind] = (
            ya[2:] - self.X_fuel[:-1]
        )

        # Fix the flame position
        residuals[self.nind] = (
            T_left - (self.T_fuel + self.delta_T)
        )

        # Upstream asymptotic conductive heat flux
        residuals[self.nind + 1] = (
            q_left + m * (h_left - self.h_fuel)
        )

        # Downstream zero conductive heat flux
        residuals[self.nind + 2] = yb[1]

        return residuals

    def initial_guess(self, x, m_guess, T_ash_guess,
                      X_ash_guess, width):
        """
        Construct a smooth initial flame profile.
        """

        # Smooth transition from fuel to ashes
        f = 0.5 * (1.0 + np.tanh(x / width))

        T = (
            self.T_fuel
            + (T_ash_guess - self.T_fuel) * f
        )

        X = (
            self.X_fuel[:-1, None] * (1.0 - f)
            + np.asarray(X_ash_guess[:-1])[:, None] * f
        )

        # Estimate q from the temperature gradient
        q = np.empty_like(T)

        for j in range(x.size):

            Xj = self.full_composition(X[:, j])

            rho, _, _, _ = eos(self.p0, T[j], Xj)

            kappa = conductivity(rho, T[j], Xj)

            q[j] = -kappa * np.gradient(T, x)[j]

        return np.vstack((T, q, X))

    def solve(self, m_guess, T_ash_guess,
              X_ash_guess, width, npts=500):

        x = np.linspace(-self.L, self.L, npts)

        y_guess = self.initial_guess(
            x, m_guess, T_ash_guess,
            X_ash_guess, width
        )

        sol = solve_bvp(
            self.rhs,
            self.bc,
            x,
            y_guess,
            p=np.array([m_guess]),
            tol=1.e-5,
            bc_tol=1.e-5,
            max_nodes=20000,
            verbose=2
        )

        return sol
```
